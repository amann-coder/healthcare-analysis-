```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
print("libraries import succesfult")
```

```python
df=pd.read_csv('test_healthcare_data.csv')
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['age'],kde=True,color='skyblue',bins=20)
plt.title('Age distribution of tested patients ', fontsize=14)
plt.xlabel('age(years)')
plt.ylabel('count ',fontsize=12)
plt.show()
```

```python
print(df.columns)
```

```python
sns.scatterplot(data=df,
                x='glucose',
                y='cholesterol',
                s=70,
                marker='*',
                color="#6D66EA",
                alpha=0.8)
plt.title("patient glucos  vs cholestrol")
plt.xlabel("glicose ")
plt.ylabel("cholestrol")
plt.show()
```

```python
plt.figure(figsize=(10,8))
correlation_matrix=df.corr()
sns.heatmap(correlation_matrix,annot=True,cmap='coolwarm',fmt=".2f",linewidths=6)
plt.title("correlation marix of health matrix")
plt.show()
```
